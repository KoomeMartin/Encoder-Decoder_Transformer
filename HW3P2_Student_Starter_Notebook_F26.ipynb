## The model in 8 simple steps

1. **Features.** Each utterance arrives as 80 log-mel filterbank values per 10 ms frame: an `(80, T)` array.
2. **Normalize, pad, augment** (`ASRDataset`).
   - Every frequency bin is standardized with training-set statistics (global MVN).
   - Utterances are zero-padded into a batch, and their true lengths are kept.
   - During training, SpecAugment blanks random frequency bands and time spans, so the model can't rely on any single one.
3. **Speech embedding** (`SpeechEmbedding`). Two strided 3×3 convolutions shorten the time axis 4× (≈1270 → 318 frames for an average utterance). They also turn each frame into a `d_model = 256` vector. Attention cost grows with T², so every encoder layer becomes ~16× cheaper.
4. **Encoder** (12 × `SelfAttentionEncoderLayer`). Each layer is pre-norm: `x + SelfAttn(RMSNorm(x))`, then `x + SwiGLU(RMSNorm(x))`.
   - Self-attention is **bidirectional**: only padding is masked, so every frame uses context before and after it.
   - Position enters through **RoPE**, which rotates queries and keys so attention depends on relative distance.
5. **CTC head.** `Linear + LogSoftmax` on the encoder output gives a CTC loss. This auxiliary loss forces the encoder *alone* to line audio up with text left to right. That makes the decoder's job much easier early in training.
6. **Decoder** (6 × `CrossAttentionDecoderLayer`). It reads the transcript so far (`[SOS] THE CAT ...`) through three sublayers:
   - **causal self-attention:** it sees only earlier tokens;
   - **cross-attention:** queries from the text, keys and values from the encoder, so it decides which audio frames to listen to;
   - **SwiGLU feed-forward.**
7. **Output and loss.**
   - A final norm and a linear layer give logits over the 5,000-token BPE vocabulary at every position.
   - Training compares them with the transcript shifted by one position (*teacher forcing*) using cross-entropy.
   - Total loss = `CE + 0.3 × CTC`.
8. **Decoding.** The encoder runs once. The decoder then generates one token at a time from `[SOS]` until `[EOS]`:
   - **greedy:** always take the most likely token;
   - **beam search:** keep the best *k* partial transcripts.

   Which checkpoint and which beam width to use are both chosen on the dev set.

## What this setup incorporates to maximize performance

| Choice | YAML setting | Why |
|---|---|---|
| Subword tokens | `token_type: 5k` | Short targets (≈4 characters per token): a faster decoder and fewer decoding steps (writeup §6.1.1) |
| 4× time reduction | `time_reduction: 4` | ~16× cheaper attention; fits T4 memory and time (§6.1.3) |
| Deep encoder, lighter decoder | 12 enc / 6 dec, `d_model: 256`, 23.9M params | Modelling the audio is the hard part; stays under the 30M limit |
| Modern pre-norm blocks | `rope`, `rmsnorm`, `swiglu`, `qk_norm: True` | QK-norm keeps attention logits bounded, which steadies fp16 training |
| Joint CTC + attention loss | `ctc_weight: 0.3` | Faster, more stable audio–text alignment |
| Label smoothing | `label_smoothing: 0.1` | A less over-confident decoder, which also helps beam search |
| SpecAugment | 2 freq masks ≤ 20 bins, 4 time masks ≤ 50 frames | The main regularizer for 100 hours of speech |
| Dropout, weight tying | `dropout: 0.1`, `weight_tying: True` | Regularization; tying shares the embedding and output matrices (−1.3M params) |
| AdamW + warmup + cosine | `lr: 1e-3`, 4 warmup epochs, cosine to 1e-6 | Transformers need warmup; cosine decay gives the best final epochs |
| Accumulation + clipping | batch 16 × 2, `grad_clip_norm: 1.0` | Effective batch 32 within T4 memory; clipping prevents fp16 blow-ups (§6.3, §6.4) |
| Mixed precision | built into the trainer | T4 tensor cores, ≈2–3× faster |
| Dev-set selection | `inference.checkpoint: auto`, `beam_width: auto` | Dev (never trained on) picks best vs last weights and greedy vs beam |
| Optional curriculum | `training.strategy: progressive` | Writeup §6.2 progressive training; off by default |

````python
# Helpers used later in this notebook (plain Python around the library, no model logic)
import glob, random, datetime
import numpy as np

if device == 'cuda':
    props = torch.cuda.get_device_properties(0)
    print(f"GPU: {props.name} | {props.total_memory / 1e9:.1f} GB")
else:
    print("No GPU found: fine for the tests, far too slow for training")

def set_seed(seed):
    """Make runs repeatable (data order, SpecAugment masks, initialization)."""
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)

def resolve_data_root(root):
    """'auto' -> the first hw3p2_data folder found for Kaggle, Colab or PSC, else the handout subset."""
    if root and str(root).lower() != 'auto':
        return root
    candidates = []
    for depth in range(1, 7):
        candidates += sorted(glob.glob('/kaggle/input/' + '*/' * (depth - 1) + 'hw3p2_data'))
    candidates += ['/content/hw3p2-data/hw3p2_data', '/local/hw3p2-data/hw3p2_data', 'hw3_data_subset/hw3p2_data']
    for path in candidates:
        if 'hw3_data_subset' in path and not path.startswith('hw3_data_subset'):
            continue  # a copy of the handout subset (e.g. inside an attached notebook output), not the dataset
        if os.path.isdir(os.path.join(path, 'train-clean-100')):
            return path
    raise FileNotFoundError("Could not find hw3p2_data. Set data.root in the config to its path.")

def make_run_name(config):
    """'auto' -> e.g. mkoome-5k-12enc6dec-d256-cold_start-1019-1530."""
    name = config['training'].get('run_name', 'auto')
    if name and str(name).lower() != 'auto':
        return name
    m = config['model']
    stamp = datetime.datetime.now().strftime('%m%d-%H%M')
    return (f"{config['Name']}-{config['tokenization']['token_type']}-"
            f"{m['num_encoder_layers']}enc{m['num_decoder_layers']}dec-d{m['d_model']}-"
            f"{config['training'].get('strategy', 'cold_start')}-{stamp}")

def fetch_wandb_checkpoint(run_path, filename='checkpoint-last-epoch-model.pth', out_dir='resume'):
    """Download a checkpoint the trainer uploaded to a W&B run ('entity/project/run_id')."""
    run = wandb.Api().run(run_path)
    for f in run.files():
        if f.name.endswith(filename):
            f.download(root=out_dir, replace=True)
            return os.path.join(out_dir, f.name)
    raise FileNotFoundError(f"{filename} not found in W&B run {run_path}")

def recognize_and_score(trainer, loader, beam_width, max_length, name, config):
    """Decode a split that has transcripts and return (metrics, results)."""
    rc = {'num_batches': None, 'beam_width': beam_width,
          'temperature': config['inference'].get('temperature', 1.0),
          'repeat_penalty': config['inference'].get('repeat_penalty', 1.0)}
    results = trainer.recognize(loader, rc, config_name=name, max_length=max_length)
    metrics = trainer._calculate_asr_metrics([r['target'] for r in results], [r['generated'] for r in results])
    return metrics, results

def show_latest_attention(trainer):
    """Display the newest decoder self- and cross-attention plots saved by the trainer."""
    from IPython.display import Image, display
    for kind in ['decoder_self', 'decoder_cross']:
        plots = glob.glob(os.path.join(str(trainer.attn_dir), f'{kind}_attention_epoch*.png'))
        if plots:
            latest = max(plots, key=lambda p: int(p.rsplit('epoch', 1)[1].split('.')[0]))
            print(os.path.basename(latest))
            display(Image(filename=latest, width=520))
        else:
            print(f"No {kind} plots yet")
````

### What the sections control
- `tokenization`: `token_type` picks the vocabulary (`char`, `1k`, `5k`, `10k`).
- `data`:
  - `root: "auto"` finds the dataset on Kaggle, Colab or PSC.
  - `load_threads: 32` reads files in parallel, so loading the full training set takes minutes instead of hours on Kaggle.
  - `feat_dtype: float16` halves RAM use.
  - `specaug` / `specaug_conf` control augmentation.
- `model`: architecture (layers, width, heads, positional encoding, norms, regularization). The model cell asserts the 30M parameter limit.
- `training`:
  - `strategy`: `cold_start` or `progressive`.
  - `epochs`, gradient accumulation and clipping.
  - `time_budget_hours`: stops cleanly before a session limit.
  - W&B settings, and `resume_checkpoint` to continue a run.
- `progressive`: the stages used only when `training.strategy: "progressive"` (writeup §6.2).
- `loss`: label smoothing and the CTC weight. Total loss = CE + `ctc_weight` × CTC.
- `optimizer` / `scheduler`: AdamW with linear warmup, then cosine decay. `cosine.T_max: "auto"` = total epochs − warmup epochs.
- `inference`: which checkpoint and beam width to use for the test set (`"auto"` = chosen on the dev set).

`BE CAREFUL` with numbers: YAML reads `1e-4` as a string, but `1.0e-4` as a float.

## Calculate Max Lengths
Calculating the maximum transcript length across your dataset is a crucial step when working with certain transformer models.
-  The positional encodings, sinusoidal or rotary, are precomputed up to a fixed maximum length.
- This maximum length is a hyperparameter that determines:
  - How long of a sequence your model can process
  - The size of your positional encoding matrix
  - Memory requirements during training and inference
- `Requirements`: For this assignment, ensure your positional encodings can accommodate at least the longest sequence in your dataset to prevent truncation.
- `NOTE`: We'll be using the same positional encoding matrix for all sequences in your dataset. Take this into account when setting your maximum length.

### Model, trainer and training strategy
`training.strategy` in the YAML picks one of two strategies:
- **`cold_start`** (default): train the full model from step 1. Warmup + cosine and gradient clipping make this stable, and it is the simplest to reason about.
- **`progressive`** (writeup §6.2): `ProgressiveTrainer` first trains a shallow model on part of the data (the `progressive.stages`). It then restores all layers, unfreezes everything and trains the full model for `training.epochs`. It can converge faster early on, but resuming is only supported once the full-model phase has started.

#### Train
Each loop pass trains one epoch, then decodes a few dev batches to report WER/CER. The loop stops when `TOTAL_EPOCHS` is reached, or earlier if the next epoch would pass `training.time_budget_hours`. In both cases the rest of the notebook still runs.

What to watch:
- `train CE` should fall steadily.
- Dev WER drops sharply once cross-attention locks onto the audio, usually within the first ~5–10 epochs.
- If the loss jumps or becomes NaN, lower `optimizer.lr`.

````python
budget_h   = config['training'].get('time_budget_hours')
start_time = globals().get('NOTEBOOK_START', time.time())

# Optional curriculum (writeup 6.2): grow the model stage by stage, only from a fresh start
if STRATEGY == 'progressive':
    if trainer.current_epoch == 0:
        trainer.progressive_train(train_loader, val_loader, PROGRESSIVE_STAGES)
    trainer.transition_to_full_training()  # restore all layers, unfreeze, full regularization

# Main phase: one epoch per call, so training can stop cleanly before the session limit
epoch_times = []
while trainer.current_epoch < TOTAL_EPOCHS:
    t0 = time.time()
    ASRTrainer.train(trainer, train_loader, val_loader, epochs=1)
    epoch_times.append(time.time() - t0)
    h = trainer.training_history[-1]
    print(f"Epoch {trainer.current_epoch}/{TOTAL_EPOCHS} | {epoch_times[-1] / 60:.1f} min | "
          f"train CE {h['train']['ce_loss']:.3f} CTC {h['train']['ctc_loss']:.3f} | "
          f"dev WER {h['val']['wer']:.2f}% CER {h['val']['cer']:.2f}% | lr {h['lr']:.2e}")
    elapsed_h = (time.time() - start_time) / 3600
    if budget_h and elapsed_h + 1.2 * max(epoch_times[-3:]) / 3600 > budget_h:
        print(f"Stopping after epoch {trainer.current_epoch}: another epoch would pass the {budget_h} h budget. "
              f"Continue later with training.resume_checkpoint.")
        break
````

#### Choose the checkpoint and beam width on the full dev set
The dev set is used only to *select* settings, never to train. With the YAML defaults:
- `inference.checkpoint: "auto"` decodes all of dev-clean greedily with the best and the last checkpoint, then keeps the lower WER.
- `inference.beam_width: "auto"` compares the widths in `inference.dev_beam_widths` and keeps the best one for the test set.

Beam search is slower (≈ beam width × greedy time) but usually lowers WER.

````python
ckpt_choice = str(config['inference'].get('checkpoint', 'auto'))
ckpt_files  = {'best': trainer.best_model_path, 'last': trainer.last_model_path}

if ckpt_choice.lower() == 'auto':
    dev_wer = {}
    for name, path in ckpt_files.items():
        if os.path.exists(path):
            trainer.load_checkpoint(path)
            dev_wer[name] = recognize_and_score(trainer, val_loader, 1, max_transcript_len, f'dev_{name}_greedy', config)[0]['wer']
            print(f"{name:>4} checkpoint: dev WER {dev_wer[name]:.2f}% (greedy)")
    ckpt_choice = min(dev_wer, key=dev_wer.get) if dev_wer else 'current'

if ckpt_choice in ckpt_files:
    trainer.load_checkpoint(ckpt_files[ckpt_choice])
elif ckpt_choice not in ('current', 'none'):
    trainer.load_checkpoint(ckpt_choice)  # an explicit path
print(f"Using the '{ckpt_choice}' weights")

bw_setting = config['inference'].get('beam_width', 1)
widths = config['inference'].get('dev_beam_widths', [1]) if str(bw_setting).lower() == 'auto' else [int(bw_setting)]
dev_metrics = {}
for bw in widths:
    metrics, dev_results = recognize_and_score(trainer, val_loader, bw, max_transcript_len, f'dev_beam{bw}', config)
    dev_metrics[bw] = metrics
    print(f"beam {bw:>2}: dev WER {metrics['wer']:.2f}% | CER {metrics['cer']:.2f}%")
SELECTED_BEAM_WIDTH = min(dev_metrics, key=lambda b: dev_metrics[b]['wer'])
print(f"\nBeam width for the test set: {SELECTED_BEAM_WIDTH}")

if trainer.use_wandb and trainer.wandb_run is not None:
    trainer.wandb_run.summary.update({f'dev_full/wer_beam{b}': m['wer'] for b, m in dev_metrics.items()})
````
